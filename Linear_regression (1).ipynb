```python
import numpy as np
import matplotlib.pyplot as plt

class linear_regression:
   def __init__(self,learning_rate=0.01,epochs=1000):
    self.lr=learning_rate
    self.epochs=epochs
    self.weight=0
    self.bias=0
   def fit(self,X,y):
     n=len(X)
     for _ in range(self.epochs):
       y_pred=self.weight*X+self.bias
       dw=(-2/n)*np.sum(X*(y-y_pred))
       db=(-2/n)*np.sum(y-y_pred)
       self.weight-=self.lr*dw
       self.bias-=self.lr*db
   def predict(self,x):
     return self.weight*x+self.bias
if __name__=="__main__":
  np.random.seed(0)
  X=np.linspace(0,10,200)
  Y=2*X+1+np.random.randn(200)*2
  model=linear_regression(learning_rate=0.01,epochs=1000)
  model.fit(X,Y)
  print(f"learned weight(slope):{model.weight:.3f}")
  print(f"learned weight(intesection):{model.bias:.3f}")

  Y_pred=model.predict(X)

  plt.scatter(X,Y,label="Data")
  plt.plot(X,Y_pred,color='red',label="Regression_Line")
  plt.legend()
  plt.xlabel('X')
  plt.ylabel("Y")
  plt.show()
```
